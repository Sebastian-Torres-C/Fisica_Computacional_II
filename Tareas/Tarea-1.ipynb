```python
"""
b) Cree un diccionario que contenga al menos un elemento de tipo int, uno de tipo
float, uno de tipo list y uno tipo tuple. Luego, imprima por separado cada uno de
los elementos almacenados en el diccionario.
"""
dic= {}
dic[1]= [1,2,3,4,5]
dic[2]= 25
dic[3]= 2.24
dic[4]= (5,43,678,0)

print('Los valores int del diccionario son:')

for u in dic.values():
    if type(u) == type(1):
        print(u)

print('Los valores float del diccionario son:')

for u in dic.values():
    if type(u) == type(6.5):
        print(u)
        
print('Los valores list del diccionario son:')

for u in dic.values():
    if type(u) == type([4,3,6]):
        print(u)

print('Los valores tuple del diccionario son:')

for u in dic.values():
    if type(u) == type((3,5,7)):
        print(u)
```
